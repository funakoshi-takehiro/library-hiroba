## 1. 基本の書き方

`ui.html(HTML, css=CSS)` の2つを渡すだけです。HTML の中で付けた `class` を CSS から指定します。

```python
ui.html("""
<div class="memo">ここが自由に書ける場所です</div>
""", css="""
.memo {
  border: 2px dashed #999;
  border-radius: 12px;
  padding: 12px 16px;
}
""")
```

## 2. テーマの色を使う

色を直接書くと、ダークモードで見えにくくなることがあります。library-hiroba が用意しているテーマ変数を `var(...)` で指定すると、ライトとダークの切り替えに自動で追従します。

| 変数 | 用途 |
|---|---|
| `--hui-accent` | ブランド色（線・塗り用） |
| `--hui-accent-ink` | ブランド色の文字用（濃いめ） |
| `--hui-accent-soft` | ブランド色のうすい背景 |
| `--hui-ok` / `--hui-warn` / `--hui-bad` | 正解・注意・まちがいの色 |
| `--hui-ok-ink` など | 同じ色の文字用 |
| `--hui-ok-soft` など | 同じ色のうすい背景 |
| `--hui-on-accent` | アクセント色で塗った上に載せる文字色 |
| `--hui-ink` / `--hui-ink-3` | 本文の色 / 補助テキストの色 |
| `--hui-paper` / `--hui-bg-2` / `--hui-line` | 紙の色 / 薄い背景 / 罫線 |
| `--hui-radius` / `--hui-shadow` | 角丸 / 影 |

```python
ui.html("""
<div class="fukidashi">こんにちは！<br>いっしょに <b>Python</b> を勉強しよう</div>
""", css="""
.fukidashi {
  display: inline-block;
  border: 2px solid var(--hui-accent);
  background: var(--hui-accent-soft);
  border-radius: var(--hui-radius);
  padding: 10px 16px;
  font-weight: 700;
}
b { color: var(--hui-accent-ink); }
""")
```

## 3. レシピ集

ここからはコピーして使える例です。文言や色を書き換えて使ってください。

### 3-1. しっぽ付きのふきだし（先生のひとこと）

```python
ui.html("""
<div class="sensei">
  <div class="sensei-name">先生</div>
  <div class="fukidashi">まずは <b>print()</b> で画面に文字を出してみよう！</div>
</div>
""", css="""
.sensei { display: flex; align-items: flex-start; gap: 12px; }
.sensei-name {
  flex: none;
  width: 46px; height: 46px;
  border-radius: 999px;
  background: var(--hui-accent);
  color: var(--hui-on-accent);
  font-size: 13px; font-weight: 700;
  display: flex; align-items: center; justify-content: center;
}
.fukidashi {
  position: relative;
  border: 1px solid var(--hui-line);
  background: var(--hui-paper);
  border-radius: var(--hui-radius);
  box-shadow: var(--hui-shadow);
  padding: 12px 16px;
}
/* 左向きのしっぽ（三角形）を CSS だけで作る */
.fukidashi::before {
  content: "";
  position: absolute;
  left: -9px; top: 16px;
  border-width: 8px 9px 8px 0;
  border-style: solid;
  border-color: transparent var(--hui-line) transparent transparent;
}
.fukidashi::after {
  content: "";
  position: absolute;
  left: -8px; top: 16px;
  border-width: 8px 9px 8px 0;
  border-style: solid;
  border-color: transparent var(--hui-paper) transparent transparent;
}
""")
```

### 3-2. 手順ステップ（番号は CSS が自動で振る）

```python
ui.html("""
<ol class="steps">
  <li>変数に数を入れる</li>
  <li>for 文でくりかえす</li>
  <li>print() で表示する</li>
</ol>
""", css="""
.steps { list-style: none; counter-reset: step; padding: 0; margin: 0; }
.steps li {
  counter-increment: step;
  display: flex; align-items: center; gap: 12px;
  border: 1px solid var(--hui-line);
  background: var(--hui-paper);
  border-radius: var(--hui-radius);
  padding: 12px 16px;
  margin-bottom: 8px;
}
.steps li::before {
  content: counter(step);          /* 1, 2, 3... を自動で入れる */
  flex: none;
  width: 28px; height: 28px;
  border-radius: 999px;
  background: var(--hui-accent);
  color: var(--hui-on-accent);
  font-weight: 700;
  display: flex; align-items: center; justify-content: center;
}
""")
```

### 3-3. 単語カードのグリッド（画面幅に合わせて折り返す）

```python
words = [("apple", "りんご"), ("book", "本"), ("cat", "ねこ"),
         ("desk", "つくえ"), ("egg", "たまご"), ("fish", "さかな")]

items = "".join(
    f'<div class="word"><div class="en">{en}</div><div class="ja">{ja}</div></div>'
    for en, ja in words
)

ui.html(f'<div class="word-grid">{items}</div>', css="""
.word-grid {
  display: grid;
  grid-template-columns: repeat(auto-fill, minmax(120px, 1fr));
  gap: 10px;
}
.word {
  border: 1px solid var(--hui-line);
  background: var(--hui-paper);
  border-radius: var(--hui-radius);
  box-shadow: var(--hui-shadow);
  padding: 12px;
  text-align: center;
}
.en { font-weight: 700; color: var(--hui-accent-ink); }
.ja { font-size: 0.9em; color: var(--hui-ink-3); }
""")
```

### 3-4. 横棒グラフ（Python の集計結果をそのまま見せる）

f-string で値を差し込めば、計算結果をそのままグラフにできます。

```python
scores = {"国語": 82, "算数": 95, "理科": 74, "社会": 88}

rows = "".join(
    f'<div class="bar-row">'
    f'  <div class="bar-label">{name}</div>'
    f'  <div class="bar-track"><div class="bar-fill" style="width: {value}%"></div></div>'
    f'  <div class="bar-value">{value}</div>'
    f'</div>'
    for name, value in scores.items()
)

ui.html(f'<div class="bars">{rows}</div>', css="""
.bars { display: flex; flex-direction: column; gap: 8px; }
.bar-row { display: flex; align-items: center; gap: 10px; }
.bar-label { flex: none; width: 3.5em; }
.bar-track {
  flex: 1;
  height: 14px;
  background: var(--hui-bg-2);
  border: 1px solid var(--hui-line);
  border-radius: 999px;
  overflow: hidden;
}
.bar-fill { height: 100%; background: var(--hui-accent); border-radius: 999px; }
.bar-value { flex: none; width: 2.5em; text-align: right; font-weight: 700; }
""")
```

## 4. CSS が効く範囲（スコープ）

`ui.html()` の `css` は、既定ではその部品の内側だけに適用されます。そのため `b { ... }` のようにざっくり書いても、他のセルや教材ページの見た目はそのまま保たれます。

下のセルで確かめてみましょう。

```python
ui.stack(
    ui.html("<p>この中の <b>太字</b> は赤くなる</p>", css="b { color: var(--hui-bad-ink); }"),
    ui.html("<p>となりの <b>太字</b> は影響を受けない</p>"),
)
```

### アニメーションを使いたいとき（`scoped=False`）

`@keyframes` のような一部の指定は、スコープの内側には書けません。その場合は `scoped=False` を指定します。CSS がページ全体に適用されるので、クラス名は他とかぶらない名前にしてください。

なお、この例では `prefers-reduced-motion` に従う書き方にしてあります。OS で「視差効果を減らす」を有効にしている場合は、アニメーションが止まります。

```python
ui.html("""
<div class="hui-demo-pulse">じゅんびできたら、はじめよう</div>
""", scoped=False, css="""
@keyframes hui-demo-pulse-anim {
  0%, 100% { transform: scale(1); }
  50%      { transform: scale(1.04); }
}
.hui-demo-pulse {
  display: inline-block;
  background: var(--hui-accent);
  color: var(--hui-on-accent);
  font-weight: 700;
  border-radius: 999px;
  padding: 10px 22px;
}
@media (prefers-reduced-motion: no-preference) {
  .hui-demo-pulse { animation: hui-demo-pulse-anim 1.6s ease-in-out infinite; }
}
""")
```

## 5. 使うときの注意

- `ui.html()` に渡した HTML はそのまま表示されます。生徒が入力した文字を入れる場合は、`ui.card()` など他の部品を使ってください。そちらは自動でエスケープされます
- PyHiroba では、表示の直前に安全チェックが入ります。`<script>` やクリック時の動作（`onclick` など）、`<iframe>` と `<form>` は取り除かれます。このノートブックのレシピはすべてチェックを通ります
- `scoped=False` はページ全体に適用されます。`@keyframes` などが必要なときに限って使い、クラス名の頭に `hui-demo-` のような目印を付けると安全です
- 色は直接書くよりテーマ変数を使うと、ダークモードでも読みやすさが保たれます
