library-hiroba は、カード・クイズ・進捗バーといった UI 部品をノートブックに表示する、教育向けのライブラリです。用意された部品のほかに、HTML と CSS を自由に書くこともできます。

[PyHiroba](https://pyhiroba.weblab.t.u-tokyo.ac.jp/) でも同じように動きます。

各セルを上から順に実行してください。部品は「セル最後の式」に置くと表示されます。

小さな言語モデルを動かす `ai` の使い方は、`demo_ai.ipynb` にまとめてあります。

```python
%pip install -q -U library-hiroba
# PyHiroba に同梱済みの場合は、このセルの実行は不要です

# 入れ直しても動きが変わらないときは、前の版がメモリに残っているのが原因です
import sys
from importlib.metadata import PackageNotFoundError, version

try:
    installed = version("library-hiroba")
except PackageNotFoundError:
    installed = None  # PyHiroba は同梱なので、配布情報を持ちません
loaded = getattr(sys.modules.get("library_hiroba"), "__version__", None)
if installed and loaded and loaded != installed:
    print(f"{loaded} が読み込まれたままです（入っているのは {installed}）。")
    print("メニューの「ランタイム」→「セッションを再起動する」を実行してから、")
    print("このセルをもう一度、上から順に実行してください。")
else:
    print("library-hiroba", installed or loaded)
```

```python
import library_hiroba
from library_hiroba import ui

library_hiroba.__version__
```

```python
# 説明カード
ui.card("今日の目標", "for文を使って、九九の表を作ってみよう！",
        footer="ヒント: range(1, 10) を2回使うよ")
```

```python
# ヒント・注意（info / success / warning / danger）
ui.alert("range(5) は 0〜4 まで。5 は入らないよ", kind="warning", title="よくあるまちがい")
```

```python
# 選択式クイズ — 選択肢を選ぶと色とマークで正誤が表示されます
ui.quiz("2の8乗はいくつ？", choices=[128, 256, 512], answer=256,
        explanation="2を8回かけると 256 になるよ。Python では 2**8 で計算できる。")
```

```python
# 答え・解説の開閉ボックス
ui.reveal("print('こんにちは') と書きます", summary="答えを見る")
```

```python
# 進捗バー
ui.progress(7, max=10, label="練習問題の進み具合")
```

```python
# 数値の強調表示 + 横並び配置
ui.columns(
    ui.stat("正答率", 85, unit="%"),
    ui.stat("れんぞく正解", 4, unit="問"),
    ui.stat("学習時間", 25, unit="分"),
)
```

```python
# バッジとテーブル
ui.stack(
    ui.badge("今週のポイント", color="green"),
    ui.table([
        {"名前": "佐藤", "得点": 90, "コメント": "よくできました"},
        {"名前": "鈴木", "得点": 85, "コメント": "おしい！"},
    ], caption="漢字テストの結果"),
)
```

```python
# セルの途中で表示したいときは ui.show()（ループ内でも使える）
for i, name in enumerate(["変数", "if文", "for文"], start=1):
    ui.show(ui.progress(i, max=3, label=f"ステップ{i}: {name}"))
```

```python
# HTML と CSS を自由に書けます。HTML はそのまま表示されます
# css は既定でこの部品の内側だけに適用されます（scoped=False でページ全体に適用）
# テーマ変数（--hui-accent など）を使うと、ダークモードにも自動で追従します
ui.html("""
<div class="fukidashi">こんにちは！<br>いっしょに <b>Python</b> を勉強しよう</div>
""", css="""
.fukidashi {
  display: inline-block;
  border: 2px solid var(--hui-accent);
  border-radius: 14px;
  padding: 10px 16px;
  background: var(--hui-accent-soft);
  font-weight: 700;
}
b { color: var(--hui-accent-ink); }
""")
```
