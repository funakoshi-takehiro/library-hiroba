# ui-hiroba デモ（Google Colab 用）

**ui-hiroba** は、Gradio 風の書き心地で HTML/CSS の UI 部品をノートブックに表示する、サーバー不要の教育向けライブラリです。

各セルを上から順に実行してください。部品は「セル最後の式」に置くと表示されます。

```python
import ui_hiroba as ui

ui.__version__
```

```python
# 説明カード
ui.card("今日の目標", "for文を使って、九九の表を作ってみよう！", icon="🎯",
        footer="ヒント: range(1, 10) を2回使うよ")
```

```python
# ヒント・注意（info / success / warning / danger）
ui.alert("range(5) は 0〜4 まで。5 は入らないよ", kind="warning", title="よくあるまちがい")
```

```python
# 選択式クイズ — 選ぶと色とマークで正誤がわかる（JavaScript 不使用・CSS のみ）
ui.quiz("2の8乗はいくつ？", choices=[128, 256, 512], answer=256,
        explanation="2を8回かけると 256 になるよ。Python では 2**8 で計算できる。")
```

```python
# 答え・解説の開閉ボックス
ui.reveal("print('こんにちは') と書きます", summary="答えを見る")
```

```python
# 進捗バー
ui.progress(7, max=10, label="練習問題の進み具合")
```

```python
# 数値の強調表示 + 横並び配置
ui.columns(
    ui.stat("正答率", 85, unit="%", icon="📈"),
    ui.stat("れんぞく正解", 4, unit="問", icon="🔥"),
    ui.stat("学習時間", 25, unit="分", icon="⏱"),
)
```

```python
# バッジとテーブル
ui.stack(
    ui.badge("今週のポイント", color="green"),
    ui.table([
        {"名前": "佐藤", "得点": 90, "コメント": "よくできました"},
        {"名前": "鈴木", "得点": 85, "コメント": "おしい！"},
    ], caption="漢字テストの結果"),
)
```

```python
# セルの途中で表示したいときは ui.show()（ループ内でも使える）
for i, name in enumerate(["変数", "if文", "for文"], start=1):
    ui.show(ui.progress(i, max=3, label=f"ステップ{i}: {name}"))
```

```python
# 自由な HTML を書きたいときの逃げ道（エスケープされない点に注意）
ui.html("<p>ここには <b>自由な HTML</b> が書けます</p>")
```
